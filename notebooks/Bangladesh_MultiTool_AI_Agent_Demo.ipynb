## 4. Query Individual DB Tools Programmatically

```python
from src.tools import InstitutionsDBTool, HospitalsDBTool, RestaurantsDBTool, WebSearchTool

inst_tool = InstitutionsDBTool()
hosp_tool = HospitalsDBTool()
rest_tool = RestaurantsDBTool()
search_tool = WebSearchTool()

print("=== Institutions Query ===")
print(inst_tool.run("How many colleges are in Dhaka?"))

print("\n=== Hospitals Query ===")
print(hosp_tool.run("Find DGHS facilities in Sylhet"))

print("\n=== Restaurants Query ===")
print(rest_tool.run("Top rated restaurants in Dhaka"))

print("\n=== Web Search Query ===")
print(search_tool.run("What is the healthcare policy in Bangladesh?"))
```

## 5. Main Agent Unified Routing Demo

```python
from src.agent import BangladeshAIAgent

# Initialize the main agent (optionally pass GOOGLE_API_KEY in environment or constructor)
agent = BangladeshAIAgent(verbose=True)

test_questions = [
    "How many hospitals are in Dhaka?",
    "What is the role of DGHS in Bangladesh?",
    "How many colleges are in Dhaka?",
    "Top rated restaurants in Dhaka",
    "What is the healthcare policy in Bangladesh?"
]

for q in test_questions:
    print("=" * 60)
    print(f"QUESTION: {q}")
    print(f"TARGET TOOL: {agent.route_query_offline(q)}")
    answer = agent.run(q)
    print("\nANSWER:")
    print(answer)
    print("=" * 60)
```
